# Find Outlier (number_of_cells)

```python
Q1 = df['number_of_cells'].quantile(0.25)
print("Q1 of No. of cells : ",Q1)
Q3 = df['number_of_cells'].quantile(0.75)
print("Q3 of No. of cells : ",Q3)

IQR = Q3 - Q1 
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
outlier = df[(df['number_of_cells'] < lower) | (df['number_of_cells'] > upper)]
print(outlier[['brand','model','number_of_cells']])
```

```python
import numpy as np

# log transform = Bade values ko chhota banana (scale compress karna)
df['number_of_cells'] = np.log(df['number_of_cells'])

# fill NaN with median
df['number_of_cells'] = df['number_of_cells'].fillna(df['number_of_cells'].median())
```

# Skewness (number_of_cells)

```python
import matplotlib.pyplot as plt

# skewness
print("Skewness:", df['number_of_cells'].skew())

# boxplot
plt.figure(figsize=(6,4))
plt.boxplot(df['number_of_cells'].dropna())
plt.title("Boxplot (Skewness Visualization) :")
plt.show()
```
